- Compute the mean **ignoring** NaN values using `np.nanmean`

- Replace all NaN values with the `np.nanmedian` of the array

---

````python
# create array with NaN values
arr = np.array([1.0, np.nan, 3.0, np.nan, 5.0, np.nan, 7.0])

# mean ignoring NaN values
mean_value = np.nanmean(arr)
print("Mean:", mean_value)

# sum ignoring NaN values
sum_value = np.nansum(arr)
print("Sum:", sum_value)

# find median ignoring NaN
median_value = np.nanmedian(arr)

# replace NaN values with median
arr = np.where(np.isnan(arr), median_value, arr)

# print updated array
print("Updated Array:", arr)
````

- Subtract the mean
- Divide by the standard deviation

---

````python
# create array
arr = np.array([10, 20, 30, 40, 50])

# calculate mean
mean = np.mean(arr)

# calculate standard deviation
std = np.std(arr)

# calculate z-score
z_score = (arr - mean) / std

# print z-score array
print("Z-Score:", z_score)

# verify mean is close to 0
print("Mean of Z-score:", np.mean(z_score))

# verify std is close to 1
print("Std of Z-score:", np.std(z_score))
````
